```python

import requests
import gradio as gr

# Replace with your actual API key from ExchangeRate-API.com
# You will need to sign up for a free API key at https://app.exchangerate-api.com/sign-up
API_KEY = "bc09244e6fe05eda633b2a1d"
BASE_URL = f"https://v6.exchangerate-api.com/v6/{API_KEY}/latest/"

def get_exchange_rates(base_currency):
    """Fetches the latest exchange rates for a given base currency."""
    try:
        response = requests.get(f"{BASE_URL}{base_currency}")
        response.raise_for_status()  # Raise an exception for HTTP errors
        data = response.json()
        if data["result"] == "success":
            return data["conversion_rates"]
        else:
            return None
    except requests.exceptions.RequestException as e:
        print(f"API request failed: {e}")
        return None

def convert_currency(amount, base_currency, target_currency):
    """Converts an amount from base currency to target currency."""
    if not amount or amount <= 0:
        return "Please enter a valid amount."

    rates = get_exchange_rates(base_currency)
    if rates is None:
        return "Error fetching exchange rates. Please check your API key and base currency."

    if target_currency not in rates:
        return f"Invalid target currency code: {target_currency}"

    converted_amount = amount * rates[target_currency]
    return f"{converted_amount:.2f} {target_currency}"

# Get a list of common currency codes for the dropdowns
# In a real application, you might fetch this from an API endpoint
# For this assignment, we'll use a predefined list for simplicity
COMMON_CURRENCIES = [
    "USD", "EUR", "GBP", "JPY", "CAD", "AUD", "CHF", "CNY", "INR", "PKR"
]

# Gradio Interface
iface = gr.Interface(
    fn=convert_currency,
    inputs=[
        gr.Number(label="Amount"),
        gr.Dropdown(choices=COMMON_CURRENCIES, label="Base Currency", value="USD"),
        gr.Dropdown(choices=COMMON_CURRENCIES, label="Target Currency", value="PKR")
    ],
    outputs="text",
    title="Currency Converter",
    description="Convert currencies using real-time exchange rates. Please ensure you have a valid API key from ExchangeRate-API.com."
)

if __name__ == "__main__":
    iface.launch(share=True)
```
