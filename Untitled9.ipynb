```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
import yfinance as yf
import warnings, os, random, math
warnings.filterwarnings("ignore")
tf.random.set_seed(42)
np.random.seed(42)
```

```python

def plot_series(true, pred=None, title="", zoom=None):
    plt.figure(figsize=(14,4))
    plt.plot(true, label="Actual", linewidth=1)
    if pred is not None:
        plt.plot(pred, label="Prediction", linewidth=1)
    plt.title(title); plt.legend(); plt.grid(alpha=.3)
    if zoom:
        plt.xlim(zoom)
    plt.show()
```

```python

ticker = "AAPL"
start, end = "2018-01-01", "2024-01-01"
df = yf.download(ticker, start=start, end=end, progress=False)
if df.empty:
    raise RuntimeError("Yahoo download failed – check internet or ticker.")
close = df["Close"].dropna()
print(f"✅ Loaded {len(close)} AAPL closing prices.")
```

```python

plt.figure(figsize=(12,4))
plt.plot(close[::-1], color="tab:blue")
plt.title("AAPL closing prices (oldest → newest)")
plt.show()
```

```python

prices = close.values.reshape(-1,1).astype("float32")
print("Shape ready for scaling:", prices.shape)
```

```python

x0 = tf.constant([[1.0, 0.5]], dtype=tf.float32)
x1 = tf.constant([[0.8, -0.3]], dtype=tf.float32)
Wx = tf.Variable(tf.random.normal([2,4]))
Wh = tf.Variable(tf.random.normal([4,4]))
b  = tf.Variable(tf.zeros([4]))

h0 = tf.tanh(x0 @ Wx + b)
h1 = tf.tanh(x1 @ Wx + h0 @ Wh + b)
print("h0:", h0.numpy())
print("h1:", h1.numpy())
```

```python

scaler = MinMaxScaler()
scaled = scaler.fit_transform(prices)

def make_windows(series, lookback):
    X, y = [], []
    for i in range(lookback, len(series)):
        X.append(series[i-lookback:i, 0])
        y.append(series[i, 0])
    return np.array(X), np.array(y)

LOOKBACK = 10
X, y = make_windows(scaled, LOOKBACK)
print("Window shapes:", X.shape, y.shape)


split = int(len(X)*0.8)
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

X_train = X_train.reshape((*X_train.shape, 1))
X_test  = X_test.reshape((*X_test.shape, 1))
```

```python

rnn_model = models.Sequential([
    layers.Input(shape=(LOOKBACK,1)),
    layers.SimpleRNN(32),
    layers.Dense(1)
])
rnn_model.compile(optimizer="adam", loss="mse")
rnn_model.summary()
```

```python

cb = tf.keras.callbacks.EarlyStopping(patience=5, restore_best_weights=True)
history = rnn_model.fit(X_train, y_train,
                        validation_split=0.15,
                        epochs=50, batch_size=32,
                        callbacks=[cb], verbose=0)
print("🏁 RNN training done – best val-loss:", min(history.history["val_loss"]))
```

```python

pred_scaled = rnn_model.predict(X_test, verbose=0)
pred_rnn = scaler.inverse_transform(pred_scaled.reshape(-1,1)).flatten()
true_rnn = scaler.inverse_transform(y_test.reshape(-1,1)).flatten()
rmse_rnn = math.sqrt(mean_squared_error(true_rnn, pred_rnn))
print(f"RMSE PlainRNN: ${rmse_rnn:.2f}")
plot_series(true_rnn, pred_rnn, "AAPL – Plain RNN")
```

```python

cnn_model = models.Sequential([
    layers.Input(shape=(LOOKBACK,1)),
    layers.Conv1D(filters=5, kernel_size=5, activation="relu"),
    layers.MaxPooling1D(pool_size=5),
    layers.Flatten(),
    layers.Dense(1)
])
cnn_model.compile(optimizer="adam", loss="mse")
cnn_model.summary()
```

```python
cnn_model.fit(X_train, y_train,
              validation_split=0.15,
              epochs=50, batch_size=32,
              callbacks=[cb], verbose=0)
```

```python
pred_cnn_scaled = cnn_model.predict(X_test, verbose=0)
pred_cnn = scaler.inverse_transform(pred_cnn_scaled.reshape(-1,1)).flatten()
rmse_cnn = math.sqrt(mean_squared_error(true_rnn, pred_cnn))  # same true labels
print(f"RMSE 1D-CNN: ${rmse_cnn:.2f}")
plot_series(true_rnn, pred_cnn, "AAPL – 1D CNN")
```

```python

hybrid = models.Sequential([
    layers.Input(shape=(LOOKBACK,1)),
    layers.Conv1D(filters=5, kernel_size=3, activation="relu"),
    layers.SimpleRNN(32),
    layers.Dense(1)
])
hybrid.compile(optimizer="adam", loss="mse")
hybrid.summary()
```

```python
hybrid.fit(X_train, y_train,
           validation_split=0.15,
           epochs=50, batch_size=32,
           callbacks=[cb], verbose=0)
```

```python
pred_hyb_scaled = hybrid.predict(X_test, verbose=0)
pred_hyb = scaler.inverse_transform(pred_hyb_scaled.reshape(-1,1)).flatten()
rmse_hyb = math.sqrt(mean_squared_error(true_rnn, pred_hyb))
print(f"RMSE Hybrid: ${rmse_hyb:.2f}")
plot_series(true_rnn, pred_hyb, "AAPL – Hybrid CNN+RNN")
```

```python

import pandas as pd
summary = pd.DataFrame({
    "Model": ["Plain RNN", "1D CNN", "CNN+RNN"],
    "RMPE ($)": [rmse_rnn, rmse_cnn, rmse_hyb]
}).sort_values("RMPE ($)")
print(summary.to_string(index=False))
```

```python

ticker2 = "IBM"
df2 = yf.download(ticker2, start="2015-01-01", end="2024-01-01", progress=False)
close2 = df2["Close"].dropna()
prices2 = close2.values.reshape(-1,1).astype("float32")

scaler2 = MinMaxScaler()
scaled2 = scaler2.fit_transform(prices2)

LOOKBACK_IBM = 15
X2, y2 = make_windows(scaled2, LOOKBACK_IBM)
split2 = int(len(X2)*0.75)
X2_train, X2_test = X2[:split2].reshape(-1,LOOKBACK_IBM,1), X2[split2:].reshape(-1,LOOKBACK_IBM,1)
y2_train, y2_test = y2[:split2], y2[split2:]
```

```python
ibm_model = models.Sequential([
    layers.Input(shape=(LOOKBACK_IBM,1)),
    layers.Conv1D(8, 3, activation="relu"),
    layers.SimpleRNN(32, dropout=0.15, recurrent_dropout=0.15),
    layers.Dense(1)
])
ibm_model.compile(optimizer="adam", loss="mse")
ibm_model.summary()
```

```python
ibm_model.fit(X2_train, y2_train,
              validation_split=0.15,
              epochs=60, batch_size=32,
              callbacks=[cb], verbose=0)
```

```python
pred_ibm_scaled = ibm_model.predict(X2_test, verbose=0)
pred_ibm = scaler2.inverse_transform(pred_ibm_scaled.reshape(-1,1)).flatten()
true_ibm = scaler2.inverse_transform(y2_test.reshape(-1,1)).flatten()
rmse_ibm = math.sqrt(mean_squared_error(true_ibm, pred_ibm))
print(f"IBM Hybrid (w/ dropout) RMSE: ${rmse_ibm:.2f}")

plot_series(true_ibm, pred_ibm, "IBM – Hybrid CNN+RNN (full)")
plot_series(true_ibm, pred_ibm, "IBM – Hybrid CNN+RNN (zoom)", zoom=(0,200))
```

```python

print("""
Comparison (AAPL, 2018-2024, lookback=10):
- Plain RNN     : simple, fast, but error accumulates → RMSE $"""f"{rmse_rnn:.2f}" """
- 1D CNN        : catches local patterns, no recurrence → RMSE $"""f"{rmse_cnn:.2f}" """
- Hybrid CNN+RNN: best of both worlds               → RMSE $"""f"{rmse_hyb:.2f}" """

Observations:
1. CNN alone is surprisingly strong for short lookbacks.
2. Adding RNN after CNN smooths the prediction curve.
3. Dropout helps on smaller datasets (see IBM run).
4. All models under-estimate large sudden spikes – no fundamental data fed.
""")
```
